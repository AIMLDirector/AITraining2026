```python
import nltk 
from nltk.tokenize import word_tokenize
text= """
I am learning python, java, javascript  and AI features to work on security and SOC
we are going to use AI for security audit and to identify the hackers
"""

```

```python
sentences = [word_tokenize(text.lower())]
print(sentences)
```

```python
from gensim.models import Word2Vec

cbow_model = Word2Vec(sentences=sentences, vector_size=100, window=5, min_count=1,sg=0)
```

```python
skipgram_model = Word2Vec(sentences=sentences, vector_size=100, window=5, min_count=1,sg=1)
```

```python
print(cbow_model.wv.most_similar("learning"))
```

```python
print(skipgram_model.wv.most_similar("security"))
```

```python
print(cbow_model.wv.similarity("java", "python"))
```

```python
print(cbow_model.wv.similar_by_word("python"))  # -1 to 1  if the word is similar then it wil be close to 1 ( .7 to .9 ) 
```
